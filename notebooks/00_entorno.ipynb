# 00 · Entorno reproducible

Este notebook comprueba las versiones necesarias. Las instalaciones se hacen
desde `requirements.txt`; no se mezclan comandos de terminal con código Python.

````python
from pathlib import Path
from importlib.metadata import PackageNotFoundError, version
import sys

ROOT = next(
    ruta for ruta in [Path.cwd(), Path.cwd().parent]
    if (ruta / "requirements.txt").exists()
)

PAQUETES = [
    "guppylang", "qnexus", "numpy", "pandas",
    "networkx", "scipy", "cvxpy", "matplotlib", "seaborn",
    "shapely", "pyproj",
]

print("Raíz:", ROOT.resolve())
print("Python:", sys.version.split()[0])
for paquete in PAQUETES:
    try:
        print(f"{paquete}: {version(paquete)}")
    except PackageNotFoundError:
        print(f"{paquete}: NO INSTALADO")
````

## Instalación

En una terminal de Nexus Lab, desde la raíz del proyecto:

```bash
python -m pip install -r requirements.txt
```

Después reinicie el kernel y ejecute nuevamente este notebook.
